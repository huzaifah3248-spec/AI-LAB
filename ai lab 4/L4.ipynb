```python
sentence = input("Enter a sentence: ")
words = sentence.split()
frequency = {}
for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

for word, count in frequency.items():
    print(f"{word}: {count}")
```

```python
integer=[]
input_count = int(input("How many integers would you like to enter? "))
for _ in range(input_count):
    integer.append(int(input("Enter an integer: ")))
positive_int=[]

for num in integer:
    if num >= 0:
        positive_int.append(num)

print(positive_int)
```

```python
students= {
    'huzaifa':[100,100,89],
    'ali':[90,85,95],
    'umer':[80,85,90]
}

def show_student_avg(students):
    for student in students:
        avg = sum(students[student])/len(students[student])
        print(f"{student}: {avg:.2f}")
show_student_avg(students)
        
```

```python
random = []
input_count=int(input("How many random numbers would you like to enter? "))
for i in range(input_count):
    random.append(int(input("Enter a random number: ")))
print(f'original string: {random}')

def get_even_numbers(random):
    return [num for num in random if num % 2 == 0]
print(f'even numbers: {get_even_numbers(random)}')

```

```python
class Car:
    total_cars = 0

    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed
        Car.total_cars += 1

    def drive(self):
        print(f"The {self.brand} car is driving at {self.speed} km/h.")

car1 = Car("Toyota", 120)
car2 = Car("Honda", 140)
car3 = Car("Ford", 110)

print(f"Total cars created: {Car.total_cars}")
```

```python
class Agent:
    def __init__(self, name, energy=100):
        self.name = name
        self.energy = energy
    def act(self):
        self.energy = max(0, self.energy - 10)
        print(f"{self.name} performed an action. Energy is now {self.energy}.")
class LearningAgent(Agent):
    def learn(self, topic):
        print(f"{self.name} is studying {topic}...")
        self.act()
class DecisionAgent(Agent):
    def decide(self, options):
        chosen = options[0] if options else "None"
        print(f"{self.name} evaluated {options} and selected: {chosen}.")
        self.act()
learner = LearningAgent("AlphaLearner")
decider = DecisionAgent("BetaDecider")
learner.learn("Reinforcement Learning")
learner.learn("Neural Networks")
decider.decide(["Path A", "Path B", "Path C"])
decider.decide(["Retry", "Abort"])
print(f"{learner.name} final energy: {learner.energy}")
print(f"{decider.name} final energy: {decider.energy}")
```
