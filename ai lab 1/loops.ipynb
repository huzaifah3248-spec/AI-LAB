```python
def check(attendance,marks):
    
    if attendance > 75 and marks > 50:
        eligibility = True
    else:
        eligibility = False
        
    return eligibility

def display(a,b,c,d):
    print(f"Name: {a} \nAttendence : {b} \nmarks : {c} \nEligibility: {d}")
    print("---------------------------------------------------")
    
students = []


def input_values_a():
    name = input("enter name")
    attendance = int(input("enter attendance"))
    marks = int(input("enter marks"))      

    display(name,attendance,marks,check(attendance,marks))
    students.append((name,attendance,marks,check(attendance,marks)))

num=int(input("enter number of students"))
for i in range(num):
    input_values_a()
```

```python
def display(a, b, c, d, e, f):
    print(f"Name: {a} \nBasePay : {b} \nTax : {c} \nEligibility: {d} \nNetPay: {e} \nBonus: {f}")
    print("---------------------------------------------------")

def check_eligibility():
    while True:
        a = input("Is the employee eligible for a bonus? (yes/no): ").lower().strip()
        if a == "yes":
            eligibility = True
            break
        elif a == "no":
            eligibility = False
            break
        else:
            print("Please enter 'yes' or 'no'.")
            
    return eligibility


def get_salary(basepay=10000):
    name = input("enter name: ")
    
    basepay_input = input("enter basepay (or press Enter for default 10000): ").strip()
    
    if basepay_input != "":
        basepay = int(basepay_input)
    
    eligibility = check_eligibility()
    
    if eligibility:
        bonus = 0.1 * basepay
    else:
        bonus = 0.0
        
    tax = 0.2 * basepay
    net_salary = basepay + bonus - tax 
    
    display(name, basepay, tax, eligibility, net_salary, bonus)

num = int(input("enter number of employees: "))
for i in range(num):
    print(f"\n--- Employee {i+1} ---")
    get_salary()
```
