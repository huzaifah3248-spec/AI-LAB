```python
def simulate_thermostat(start_temp, target_temp, heat_gain, total_minutes):
    current_temp = start_temp
    compressor_on = False
    lockout_timer = 0
    compressor_minutes = 0

    turn_on_temp = target_temp + 2
    turn_off_temp = target_temp - 2

    for minute in range(1, total_minutes + 1):
        if compressor_on and current_temp <= turn_off_temp:
            compressor_on = False
            lockout_timer = 3
        elif not compressor_on and lockout_timer == 0 and current_temp >= turn_on_temp:
            compressor_on = True

        if compressor_on:
            current_temp = current_temp + (heat_gain - 3)
            compressor_minutes = compressor_minutes + 1
        else:
            current_temp = current_temp + heat_gain
            if lockout_timer > 0:
                lockout_timer = lockout_timer - 1

    return compressor_minutes
runtime = simulate_thermostat(start_temp=26, target_temp=20, heat_gain=1, total_minutes=10)
print("Scenario 1 (Compressor Runtime):", runtime, "minutes")
```

```python
def simulate_tank(start_water, max_capacity, rain_per_hour, draw_per_hour, total_hours):
    current_water = start_water
    pump_on = False
    overflow_warning_count = 0
    survived = True

    low_threshold = max_capacity * 0.20
    high_threshold = max_capacity * 0.60

    hour = 1
    while hour <= total_hours:
        if current_water < low_threshold:
            pump_on = True
        elif current_water >= high_threshold:
            pump_on = False

        added_water = rain_per_hour
        if pump_on:
            added_water = added_water + 15

        current_water = current_water + added_water - draw_per_hour
        if current_water >= max_capacity:
            overflow_warning_count = overflow_warning_count + 1
            current_water = max_capacity
        if current_water <= 0:
            current_water = 0
            survived = False
            break
        hour = hour + 1
    return survived, overflow_warning_count
survived, warnings = simulate_tank(start_water=50, max_capacity=100, rain_per_hour=5, draw_per_hour=10, total_hours=24)
print("Scenario 2 (Tank Survived):", survived, "| Overflows:", warnings)
```

```python
def process_sale(price, paid, tens, fives, ones):
    if paid < price:
        return False, tens, fives, ones

    change_needed = paid - price

    used_tens = 0
    used_fives = 0
    used_ones = 0

    temp_tens = tens
    temp_fives = fives
    temp_ones = ones

    while change_needed >= 10 and temp_tens > 0:
        change_needed = change_needed - 10
        temp_tens = temp_tens - 1
        used_tens = used_tens + 1

    while change_needed >= 5 and temp_fives > 0:
        change_needed = change_needed - 5
        temp_fives = temp_fives - 1
        used_fives = used_fives + 1

    while change_needed >= 1 and temp_ones > 0:
        change_needed = change_needed - 1
        temp_ones = temp_ones - 1
        used_ones = used_ones + 1

    if change_needed == 0:
        tens = tens - used_tens
        fives = fives - used_fives
        ones = ones - used_ones

        customer_cash = paid
        while customer_cash >= 10:
            tens = tens + 1
            customer_cash = customer_cash - 10
        while customer_cash >= 5:
            fives = fives + 1
            customer_cash = customer_cash - 5
        while customer_cash >= 1:
            ones = ones + 1
            customer_cash = customer_cash - 1

        return True, tens, fives, ones
    else:
        return False, tens, fives, ones
success, t, f, o = process_sale(price=17, paid=20, tens=5, fives=10, ones=20)
print(f"Scenario 3 (Sale Success): {success} | Drawer: {t} tens, {f} fives, {o} ones")
```

```python
def rocket_flight_summary(initial_fuel):
    altitude = 0.0
    velocity = 0.0
    fuel = initial_fuel
    seconds = 0
    peak_altitude = 0.0
    peak_second = 0

    while True:
        seconds = seconds + 1

        if fuel > 0:
            if fuel >= 2:
                fuel_burned = 2
            else:
                fuel_burned = fuel
            fuel = fuel - fuel_burned
            acceleration = (fuel_burned * 15) - 10
            velocity = velocity + acceleration
        else:
            velocity = velocity - 10
        altitude = altitude + velocity
        if altitude > peak_altitude:
            peak_altitude = altitude
            peak_second = seconds
        if altitude <= 0:
            break
    return peak_altitude, peak_second, seconds
peak_alt, peak_sec, total_duration = rocket_flight_summary(initial_fuel=20)
print(f"Scenario 4 (Rocket): Peak = {peak_alt}m at {peak_sec}s | Total Flight = {total_duration}s")
```

```python
def simulate_atm_entry(correct_pin, digit_stream):
    current_value = 0
    digit_count = 0
    failed_attempts = 0
    card_blocked = False
    access_granted = False

    for digit in digit_stream:
        if card_blocked:
            break

        current_value = (current_value * 10) + digit
        digit_count = digit_count + 1

        if digit_count == 4:
            if current_value == correct_pin:
                access_granted = True
                failed_attempts = 0
                break
            else:
                failed_attempts = failed_attempts + 1
                current_value = 0
                digit_count = 0
                if failed_attempts >= 3:
                    card_blocked = True
                    break

    return access_granted, card_blocked
inputs = [1, 2, 3, 4, 4, 8, 2, 1]
granted, blocked = simulate_atm_entry(correct_pin=4821, digit_stream=inputs)
print(f"Scenario 5 (ATM): Access Granted = {granted} | Card Blocked = {blocked}")
```
