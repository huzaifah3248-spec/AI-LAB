```python
class LibraryItem:
    def __init__(self, title, item_id, category):
        self.title = title
        self.item_id = item_id
        self.category = category
        self.status = "Available"


class Book(LibraryItem):
    def __init__(self, title, item_id, category, author, page_count):
        super().__init__(title, item_id, category)
        self.author = author
        self.page_count = page_count


class Magazine(LibraryItem):
    def __init__(self, title, item_id, category, issue_number):
        super().__init__(title, item_id, category)
        self.issue_number = issue_number


class Library:
    def __init__(self, name, branch_code, latitude, longitude):
        self.metadata = (name, branch_code, (latitude, longitude))
        self.catalog = []
        self.search_index = {}
        self.categories = set()

    def add_item(self, item_obj):
        self.catalog.append(item_obj)
        self.search_index[item_obj.item_id] = item_obj
        self.categories.add(item_obj.category)

    def checkout(self, item_id):
        if item_id in self.search_index:
            item = self.search_index[item_id]
            if item.status == "Available":
                item.status = "Checked Out"
                print(f"Checkout successful: {item.title}")
            else:
                print(f"Item already checked out: {item.title}")
        else:
            print(f"Item ID {item_id} not found in catalog.")

    def display_summary(self):
        print(f"Library: {self.metadata[0]} | Branch: {self.metadata[1]}")
        print(f"Coordinates: {self.metadata[2]}")
        print(f"Categories: {self.categories}")
        for item in self.catalog:
            print(f"[{item.status}] ID: {item.item_id} | Title: {item.title} | Category: {item.category}")

city_lib = Library("City Central Library", "LHR-01", 31.5204, 74.3587)

b1 = Book("Artificial Intelligence: A Modern Approach", 101, "AI", "Stuart Russell", 1152)
b2 = Book("Python Crash Course", 102, "Programming", "Eric Matthes", 544)
m1 = Magazine("IEEE Intelligent Systems", 201, "AI", 88)

city_lib.add_item(b1)
city_lib.add_item(b2)
city_lib.add_item(m1)

city_lib.checkout(101) 
city_lib.checkout(101)   
city_lib.checkout(999)   
print("\n--- Summary Report ---")
city_lib.display_summary()
```

```python
class Person:
    def __init__(self, name, age, id_number):
        self.name = name
        self.age = age
        self.id_number = id_number


class Doctor(Person):
    def __init__(self, name, age, id_number, specialty, years_of_experience):
        super().__init__(name, age, id_number)
        self.specialty = specialty
        self.years_of_experience = years_of_experience


class Patient(Person):
    def __init__(self, name, age, id_number, ailment, blood_type):
        super().__init__(name, age, id_number)
        self.ailment = ailment
        self.blood_type = blood_type


class Hospital:
    def __init__(self, name, total_beds, emergency_contact):
        self.hospital_config = (name, total_beds, emergency_contact)
        self.patient_queue = []
        self.staff_directory = {}
        self.departments = set()
        self.current_issues = set()

    def register_doctor(self, doctor_obj):
        self.staff_directory[doctor_obj.id_number] = doctor_obj
        self.departments.add(doctor_obj.specialty)

    def admit_patient(self, patient_obj):
        max_beds = self.hospital_config[1]
        if len(self.patient_queue) >= max_beds:
            print(f"Admission Denied: Hospital is at maximum capacity ({max_beds} beds).")
            return False
        self.patient_queue.append(patient_obj)
        self.current_issues.add(patient_obj.ailment)
        print(f"Admitted Patient: {patient_obj.name} for {patient_obj.ailment}.")
        return True

    def assign_doctor(self, patient_id, doctor_id):
        patient_obj = None
        for p in self.patient_queue:
            if p.id_number == patient_id:
                patient_obj = p
                break

        if not patient_obj:
            print(f"Patient ID {patient_id} not currently admitted.")
            return

        if doctor_id in self.staff_directory:
            doctor_obj = self.staff_directory[doctor_id]
            print(f"Dr. {doctor_obj.name} is now treating Patient {patient_obj.name} for {patient_obj.ailment}.")
        else:
            print(f"Doctor ID {doctor_id} not found in directory.")

    def run_daily_rounds(self):
        print(f"\n--- Daily Rounds for {self.hospital_config[0]} ---")
        print(f"Active Issues in Ward: {self.current_issues}")
        for idx, patient in enumerate(self.patient_queue, start=1):
            print(f"Round {idx}: Checking {patient.name} | Condition: {patient.ailment} | Blood Type: {patient.blood_type}")

hospital = Hospital("City General Hospital", 2, "+92-42-111-222-333")

doc1 = Doctor("Sarah Khan", 42, 501, "Cardiology", 15)
doc2 = Doctor("Ahmed Malik", 38, 502, "Neurology", 10)

hospital.register_doctor(doc1)
hospital.register_doctor(doc2)

p1 = Patient("Zain Ali", 21, 1001, "Hypertension", "O+")
p2 = Patient("Fatima Noor", 34, 1002, "Migraine", "A+")
p3 = Patient("Bilal Tariq", 29, 1003, "Asthma", "B+")

hospital.admit_patient(p1)
hospital.admit_patient(p2)
hospital.admit_patient(p3)

hospital.assign_doctor(1001, 501)
hospital.assign_doctor(1002, 999)

hospital.run_daily_rounds()
```

```python
class Product:
    def __init__(self, product_id, name, price, stock_level, tags=None):
        self.product_id = product_id
        self.name = name
        self.price = price
        self.stock_level = stock_level
        self.tags = set(tags) if tags else set()

    def apply_discount(self, percentage):
        self.price = self.price * (1 - (percentage / 100))

class Electronics(Product):
    def __init__(self, product_id, name, price, stock_level, warranty_period, brand, tags=None):
        super().__init__(product_id, name, price, stock_level, tags)
        self.warranty_period = warranty_period
        self.brand = brand

class Clothing(Product):
    def __init__(self, product_id, name, price, stock_level, size, material, tags=None):
        super().__init__(product_id, name, price, stock_level, tags)
        self.size = size
        self.material = material

class Store:
    def __init__(self):
        self.inventory = {}
        self.shopping_cart = []
        self.shipping_regions = ("North America", "Europe", "Asia")

    def register_product(self, product_obj):
        self.inventory[product_obj.product_id] = product_obj

    def add_to_cart(self, product_id):
        if product_id in self.inventory:
            product = self.inventory[product_id]
            if product.stock_level > 0:
                self.shopping_cart.append(product)
                product.stock_level -= 1
                print(f"Added to cart: {product.name} (Remaining Stock: {product.stock_level})")
            else:
                print(f"Item out of stock: {product.name}")
        else:
            print(f"Product ID {product_id} does not exist.")

    def calculate_total(self):
        total_amount = 0.0
        for item in self.shopping_cart:
            total_amount += item.price
        return total_amount

    def filter_by_tag(self, tag_name):
        matching_products = []
        for product in self.inventory.values():
            if tag_name in product.tags:
                matching_products.append(product)
        return matching_products

store = Store()

laptop = Electronics(1, "ThinkPad X1", 1200.0, 5, 24, "Lenovo", ["tech", "sale", "pc"])
shirt = Clothing(2, "Oxford Cotton Shirt", 45.0, 10, "M", "Cotton", ["apparel", "sale", "clothing"])
headphones = Electronics(3, "Wireless Earbuds", 80.0, 0, 12, "Sony", ["audio", "tech"])

store.register_product(laptop)
store.register_product(shirt)
store.register_product(headphones)

store.add_to_cart(1)
store.add_to_cart(2)
store.add_to_cart(3)
store.add_to_cart(99)

shirt.apply_discount(10)

total_due = store.calculate_total()
print(f"Total Cart Value: ${total_due:.2f}")

sale_items = store.filter_by_tag("sale")
print("Products tagged 'sale':", [item.name for item in sale_items])
```
